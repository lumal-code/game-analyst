```python
import cv2
import numpy as np

def make_collage(batch):
    while len(batch) != 9:
        batch.append(np.zeros_like(batch[0]))
    row1 = cv2.hconcat(batch[:3])
    row2 = cv2.hconcat(batch[3:6])
    row3 = cv2.hconcat(batch[6:])

    return cv2.vconcat([row1, row2, row3])
```
